```python
import pymupdf
```

```python
pdf_path = "../data/raw/lm7800.pdf"
doc = pymupdf.open(pdf_path)
```

```python
len(doc)
```

```python
page = doc[0]
text = page.get_text()
print(text)
```

```python
for page_number, page in enumerate(doc, start=1):
    text = page.get_text()

    print(f"\n--- Page {page_number} ---")
    print(text[:500])
```

```python
pages = []

for page_number, page in enumerate(doc, start=1):
    pages.append({
        "page": page_number,
        "text": page.get_text()
    })
```
